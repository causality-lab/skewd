# Plot example datasets

```python
from src.datasets import skewANsLSs
from sklearn.preprocessing import StandardScaler
from src.skewci import skewci
from pathlib import Path
import matplotlib.pyplot as plt
import random
```

```python
dataset_types = ["LSs", "ANs"]
distributions = ["AGN", "SN"]
skew_levels = {"AGN": ["med", "strong", "extreme"], "SN": ["med", "strong"]}
pure = True
pair_ids = list(range(1, 101))

col = "black"
ps = 15
ls = 14
fs = 19

random.seed(1)
```

```python
# 1) ANs(-0.455)
fig, axes = plt.subplots(1, 4, figsize=(22, 5))

dt = "ANs"
level = "med"

dists = ["SN", "SN", "AGN", "AGN"]
ids = random.sample(pair_ids, 2) + random.sample(pair_ids, 2)


for j in range(4):
    ax = axes[j]
    dataset = skewANsLSs(
        dataset_type=dt,
        distribution=dists[j],
        skewness_level=level,
        pure=pure,
        pair_id=ids[j],
        preprocessor=None,
        double=True,
    )
    ax.scatter(dataset.cause, dataset.effect,alpha=0.5, color=col, lw=0.5, s=ps)
    ax.set_xlabel("Cause", fontsize=fs)
    ax.set_ylabel("Effect", fontsize=fs)
    ax.tick_params(axis='both', labelsize=ls)
plt.tight_layout()
fig.savefig(f"example_plots/{dt}_{level}.pdf", format='pdf', dpi=600, facecolor='w', edgecolor='w', bbox_inches='tight')
```
