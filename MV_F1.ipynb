```python
from pathlib import Path
import pickle
N = 100

result_dict = {}
GT_dict = {}

for r in range(N):
    pid = r+1
    with open(f"data/MVskew6/MVskew{pid}.pkl", "rb") as file:
        XB = pickle.load(file)
    X, B = XB["X"], XB["B"]
    GT_dict[pid] = B
    save_file = (
            Path("experiments")
            / "results"
            / "MVskew6"
            / f"result_{pid}.pkl"
        )
    with open(save_file, "rb") as file:
            result = pickle.load(file)
    result_dict[pid] = result
```

```python
import numpy as np
def calc_metrics(gt, pred):
    TP = np.sum((gt == 1) & (pred == 1))
    FP = np.sum((gt == 0) & (pred == 1))
    FN = np.sum((gt == 1) & (pred == 0))
    precision = TP / (TP + FP) 
    recall    = TP / (TP + FN) 
    f1        = 2 * precision * recall / (precision + recall)
    return precision, recall, f1
```

```python
precision6_N = []
recall6_N = []
f16_N = []

precision6_KDE = []
recall6_KDE = []
f16_KDE = []

for j in range(1,101):
    gt = GT_dict[j]
    pred_N = result_dict[j]["best_graph_N"]
    pred_KDE = result_dict[j]["best_graph_KDE"]
    p, r, f = calc_metrics(gt, pred_N)
    precision6_N.append(p)
    recall6_N.append(r)
    f16_N.append(f)
    p, r, f = calc_metrics(gt, pred_KDE)
    precision6_KDE.append(p)
    recall6_KDE.append(r)
    f16_KDE.append(f)
```

```python
print(np.mean(precision6_N))
print(np.mean(precision6_KDE))
print(np.mean(recall6_N))
print(np.mean(recall6_KDE))
print(np.mean(f16_N))
print(np.mean(f16_KDE))
```
